```python
import requests
import cv2
import numpy as np

#ESP 32 CAM image server url
ESP32_CAM_URL = "http://<ESP32_CAM_IP>/capture"
```

```python
def capture_image_from_esp32_cam(url):
    try:
        # Send GET request to the ESP32-CAM server
        response = requests.get(url)
        
        # Check if the request was successful
        if response.status_code == 200:
            # Convert the byte content into a NumPy array
            image_array = np.frombuffer(response.content, np.uint8)
            
            # Decode the image into an OpenCV format
            image = cv2.imdecode(image_array, cv2.IMREAD_COLOR)
            
            # Save the image locally
            image_filename = 'captured_image.jpg'
            cv2.imwrite(image_filename, image)
            print(f"Image saved as {image_filename}")
            
            # Optionally display the image using OpenCV
            cv2.imshow("Captured Image", image)
            cv2.waitKey(0)  # Wait for any key to close the image window
            cv2.destroyAllWindows()
            
        else:
            print(f"Failed to connect to ESP32-CAM. Status code: {response.status_code}")
    
    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    capture_image_from_esp32_cam(ESP32_CAM_URL)
```
